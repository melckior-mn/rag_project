```python
def retrieve(question, k=3):
    """
    Retourne les k passages les plus similaires à la question.
    """

    embedding_question = model.encode(
        question,
        convert_to_tensor=True
    )

    scores = util.cos_sim(
        embedding_question,
        embeddings_documents
    )[0]

    indices = scores.argsort(descending=True)

    resultats = []

    for index in indices[:k]:
        resultats.append({
            "document": documents[index],
            "score": float(scores[index])
        })

    return resultats
```

```python
def retrieve(question, k=3, n=2):

    # Étape 1 : embedding de la question
    embedding_question = model.encode(
        question,
        convert_to_tensor=True
    )

    # Étape 2 : similarité avec tous les documents
    scores_embedding = util.cos_sim(
        embedding_question,
        embeddings_documents
    )[0]

    # Étape 3 : récupération des K meilleurs candidats
    indices = scores_embedding.argsort(descending=True)[:k]

    # Étape 4 : création des paires question-document
    paires = [
        [question, documents[index]]
        for index in indices
    ]

    # Étape 5 : reranking avec le Cross-Encoder
    scores_reranker = reranker.predict(paires)

    # Étape 6 : classement selon les scores du reranker
    classement = sorted(
        zip(indices, scores_reranker),
        key=lambda x: x[1],
        reverse=True
    )

    # Étape 7 : conservation des N meilleurs
    resultats = []

    for index, score in classement[:n]:
        resultats.append({
            "document": documents[index],
            "score": float(score)
        })

    return resultats
```

```python
def creer_chunks(phrases, taille_chunk, overlap):

    chunks = []

    pas = taille_chunk - overlap
    debut = 0

    while debut < len(phrases):

        fin = debut + taille_chunk

        chunk = " ".join(phrases[debut:fin])
        chunks.append(chunk)

        debut += pas

    return chunks
```

```python
def calculer_recall(chunks, phrases, questions_test, k):

    embeddings_chunks = model.encode(
        chunks,
        convert_to_tensor=True
    )

    nombre_trouves = 0

    for test in questions_test:

        question = test["question"]
        phrase_attendue = test["phrase_attendue"]

        embedding_question = model.encode(
            question,
            convert_to_tensor=True
        )

        scores = util.cos_sim(
            embedding_question,
            embeddings_chunks
        )[0]

        indices = scores.argsort(descending=True)[:k]

        for index in indices:

            if phrases[phrase_attendue] in chunks[index]:
                nombre_trouves += 1
                break

    return nombre_trouves / len(questions_test)
```

```python
configurations = [
    (2, 0),
    (2, 1),
    (3, 0),
    (3, 1),
    (4, 0),
    (4, 1)
]

resultats_benchmark = []

for taille_chunk, overlap in configurations:

    chunks_test = creer_chunks(
        phrases,
        taille_chunk,
        overlap
    )

    recall_1 = calculer_recall(
        chunks_test,
        phrases,
        questions_test,
        k=1
    )

    recall_3 = calculer_recall(
        chunks_test,
        phrases,
        questions_test,
        k=3
    )

    resultats_benchmark.append({
        "taille_chunk": taille_chunk,
        "overlap": overlap,
        "nombre_chunks": len(chunks_test),
        "Recall@1": recall_1,
        "Recall@3": recall_3
    })
```

```python
def meilleur_score(question, chunks):

    embedding_question = model.encode(
        question,
        convert_to_tensor=True
    )

    embeddings_chunks = model.encode(
        chunks,
        convert_to_tensor=True
    )

    scores = util.cos_sim(
        embedding_question,
        embeddings_chunks
    )[0]

    return float(scores.max())


print("=== QUESTIONS PERTINENTES ===")

for question in questions_pertinentes:

    score = meilleur_score(question, chunks_3)

    print(f"{score:.4f} → {question}")


print("\n=== QUESTIONS ABSENTES ===")

for question in questions_absentes:

    score = meilleur_score(question, chunks_3)

    print(f"{score:.4f} → {question}")
```

Pour chaque question, nous allons prendre son meilleur candidat obtenu par embedding, puis demander au Cross-Encoder :

« Est-ce que cette question et ce passage sont réellement pertinents ensemble ?

🧠 Pourquoi deux scores ?

Parce qu'ils ne mesurent pas exactement la même chose.

Embedding :

« Le sens de la question est-il proche du sens de ce chunk ? »

Cross-Encoder :

« En regardant directement la question ET le chunk ensemble, quelle pertinence leur relation présente-t-elle ? »

```python
question = "Qu'est-ce que l'entropie ?"

# 1. Embedding de la question
embedding_question = model.encode(
    question,
    convert_to_tensor=True
)

# 2. Embeddings des chunks
embeddings_chunks = model.encode(
    chunks_3,
    convert_to_tensor=True
)

# 3. Similarité entre la question et tous les chunks
scores_embedding = util.cos_sim(
    embedding_question,
    embeddings_chunks
)[0]

# 4. Trouver le meilleur chunk
index_meilleur = scores_embedding.argmax()

meilleur_chunk = chunks_3[index_meilleur]

print("Meilleur score embedding :",
      float(scores_embedding[index_meilleur]))

print("\nChunk sélectionné :")
print(meilleur_chunk)
```

```python
score_reranker = reranker.predict(
    [[question, meilleur_chunk]]
)

print("Score Cross-Encoder :",
      float(score_reranker[0]))
```

```python
def score_cross_encoder(question, chunks):

    # 1. Transformer la question en embedding
    embedding_question = model.encode(
        question,
        convert_to_tensor=True
    )

    # 2. Transformer les chunks en embeddings
    embeddings_chunks = model.encode(
        chunks,
        convert_to_tensor=True
    )

    # 3. Calculer la similarité avec tous les chunks
    scores_embedding = util.cos_sim(
        embedding_question,
        embeddings_chunks
    )[0]

    # 4. Trouver le meilleur chunk
    index_meilleur = scores_embedding.argmax()

    # 5. Construire la paire question + meilleur chunk
    paire = [
        question,
        chunks[index_meilleur]
    ]

    # 6. Faire évaluer cette paire par le Cross-Encoder
    score_reranker = reranker.predict([paire])[0]

    return float(score_reranker), chunks[index_meilleur]
```

```python
def rang_bon_chunk(question, phrase_attendue, chunks, phrases):

    # Embedding de la question
    embedding_question = model.encode(
        question,
        convert_to_tensor=True
    )

    # Embeddings des chunks
    embeddings_chunks = model.encode(
        chunks,
        convert_to_tensor=True
    )

    # Similarités
    scores = util.cos_sim(
        embedding_question,
        embeddings_chunks
    )[0]

    # Classement du plus similaire au moins similaire
    indices = scores.argsort(descending=True)

    # Phrase qui doit être retrouvée
    phrase = phrases[phrase_attendue]

    # Chercher le chunk contenant cette phrase
    for rang, index in enumerate(indices, start=1):

        if phrase in chunks[index]:
            return rang

    # Le bon contenu n'existe dans aucun chunk
    return None
```

```python
def retrieve(question, chunks, k=5, n=2):

    # 1. Encoder la question
    embedding_question = model.encode(
        question,
        convert_to_tensor=True
    )

    # 2. Encoder les chunks
    embeddings_chunks = model.encode(
        chunks,
        convert_to_tensor=True
    )

    # 3. Similarité question ↔ chunks
    scores_embedding = util.cos_sim(
        embedding_question,
        embeddings_chunks
    )[0]

    # 4. Récupérer les K meilleurs candidats
    indices = scores_embedding.argsort(
        descending=True
    )[:k]

    # 5. Préparer les paires pour le Cross-Encoder
    paires = [
        [question, chunks[index]]
        for index in indices
    ]

    # 6. Reranking
    scores_cross = reranker.predict(paires)

    # 7. Trier selon le Cross-Encoder
    classement = sorted(
        zip(indices, scores_cross),
        key=lambda x: x[1],
        reverse=True
    )

    # 8. Garder les N meilleurs
    resultats = []

    for index, score in classement[:n]:

        resultats.append({
            "chunk": chunks[index],
            "score": float(score)
        })

    return resultats
```

```python
def retrieve(
    question,
    chunks,
    embeddings_chunks,
    k=3,
    n=2,
    seuil=0
):

    # 1. Encoder la question
    embedding_question = model.encode(
        question,
        convert_to_tensor=True
    )

    # 2. Calculer la similarité avec les chunks
    scores_embedding = util.cos_sim(
        embedding_question,
        embeddings_chunks
    )[0]

    # 3. Sélectionner les K meilleurs candidats
    indices = scores_embedding.argsort(
        descending=True
    )[:k]

    # 4. Préparer les paires question + chunk
    paires = [
        [question, chunks[index]]
        for index in indices
    ]

    # 5. Reranking avec le Cross-Encoder
    scores_cross = reranker.predict(paires)

    # 6. Associer chaque chunk à son score
    classement = sorted(
        zip(indices, scores_cross),
        key=lambda x: x[1],
        reverse=True
    )

    # 7. Meilleur score
    meilleur_score = float(
        classement[0][1]
    )

    # 8. Vérifier si le contexte est suffisant
    if meilleur_score < seuil:

        return {
            "status": "insufficient_context",
            "score": meilleur_score,
            "resultats": []
        }

    # 9. Si le contexte est suffisant,
    #    récupérer les N meilleurs chunks
    resultats = []

    for index, score in classement[:n]:

        resultats.append({
            "chunk": chunks[index],
            "score": float(score)
        })

    return {
        "status": "ok",
        "score": meilleur_score,
        "resultats": resultats
    }
```

```python
import re

chunks_metadata = []

chunk_size = 3
overlap = 1

chunk_id_global = 0

for document in documents:

    texte = document["text"]
    source = document["source"]

    # Découpage en phrases
    phrases = re.split(r'(?<=[.!?])\s+', texte.strip())

    # Création des chunks
    pas = chunk_size - overlap

    for i in range(0, len(phrases), pas):

        chunk = " ".join(
            phrases[i:i + chunk_size]
        )

        if not chunk:
            continue

        chunks_metadata.append({
            "text": chunk,
            "source": source,
            "chunk_id": chunk_id_global
        })

        chunk_id_global += 1


print("Nombre total de chunks :", len(chunks_metadata))

for chunk in chunks_metadata:
    print("\n---")
    print("SOURCE :", chunk["source"])
    print("CHUNK ID :", chunk["chunk_id"])
    print("TEXTE :", chunk["text"])
```
